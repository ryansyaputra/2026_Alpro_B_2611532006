# BAB II
# PEMBAHASAN

## 2.1 Dasar Teori


1. Python.
Python merupakan bahasa pemrograman yang digunakan untuk membuat berbagai macam program. Python memiliki sintaks yang relatif sederhana sehingga dapat digunakan untuk mempelajari dasar-dasar pemrograman.
2. Visual Studio Code.
Visual Studio Code (VS Code) merupakan text editor yang digunakan untuk menulis dan mengelola kode program. VS Code mendukung berbagai bahasa pemrograman, termasuk Python, sehingga dapat digunakan dalam proses pengembangan program.
3. GitHub.
GitHub merupakan platform yang digunakan untuk menyimpan dan mengelola kode program secara online. GitHub dapat digunakan sebagai tempat penyimpanan repository sehingga kode dalam suatu project dapat dikelola dengan lebih terorganisir.

## 2.2 Kode Program

### 2.2.1 Program "AlproB_2611532006"


```python
"""
komentar
dua baris atau lebih
pertemuan 1 praktikum alpro
"""
print("Hello Word")
print()
print("Program ini menghasilkan")
print("Empat baris output")
print("/ \\ // \\\ /// \\\\\\")
print("program ini" + "dibuat oleh a\n"
+ "Ryan Syaputra dan 2611532006")
```

**Penjelasan :**
1. Bagian komentar pada awal program digunakan untuk memberikan keterangan mengenai kode dan pertemuan praktikum
2. Fungsi print() digunakan untuk menampilkan teks atau menghasilkan garis kosong pada output
3. Penggunaan tanda + pada kode digunakan untuk menggabungkan beberapa teks menjadi satu output, sedangkan \n digunakan untuk membuat baris baru

## 2.3 Tahapan Commit dan Push ke GitHub




**Penjelasan :**
1. Membuat Repository di GitHub
2. Menyiapkan File Project
3. Membuka Folder di VS Code
4. Melakukan Commit dengan code
   ` git init `
   ` git add `
   ` git commit -m `
5. Menghubungkan dengan Repository GitHub
   ` git remote add origin `
6. Melakukan Push 
   ` git push -u origin main `
